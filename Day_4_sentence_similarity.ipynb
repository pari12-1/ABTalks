```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity, cosine_distances
import numpy as np

print("Libraries imported successfully!")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model loaded successfully!")
```

```python
sentences = [
    "I love playing football with my friends.",
    "I enjoy playing soccer with my classmates.",
    "Football is my favorite sport.",
    "Python is a popular programming language.",
    "I like cooking delicious food at home."
]

print("Sentences:")

for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")
```

```python
embeddings = model.encode(sentences)

print(" embeddings generated successfully !")
print(" Shape of embeddings :", embeddings.shape)
```

```python
print("First sentence:")
print(sentences[0])

print("\nFirst 10 values of its embedding:")
print(embeddings[0][:10])
```

```python
sentence1_embedding = embeddings[0].reshape(1, -1)
sentence2_embedding = embeddings[1].reshape(1, -1)

similarity = cosine_similarity(
    sentence1_embedding,
    sentence2_embedding
)[0][0]

distance = cosine_distances(
    sentence1_embedding,
    sentence2_embedding
)[0][0]

print("Sentence 1:")
print(sentences[0])

print("\nSentence 2:")
print(sentences[1])

print("\nCosine Similarity:", round(similarity, 4))
print("Cosine Distance:", round(distance, 4))
```

```python
similarity = cosine_similarity(
    embeddings[0].reshape(1, -1),
    embeddings[3].reshape(1, -1)
)[0][0]

distance = cosine_distances(
    embeddings[0].reshape(1, -1),
    embeddings[3].reshape(1, -1)
)[0][0]

print("Sentence 1:")
print(sentences[0])

print("\nSentence 4:")
print(sentences[3])

print("\nCosine Similarity:", round(similarity, 4))
print("Cosine Distance:", round(distance, 4))
```

```python
print("SIMILARITY BETWEEN ALL SENTENCE PAIRS")
print("=" * 70)

pair_results = []

for i in range(len(sentences)):

    for j in range(i + 1, len(sentences)):

        similarity = cosine_similarity(
            embeddings[i].reshape(1, -1),
            embeddings[j].reshape(1, -1)
        )[0][0]

        distance = cosine_distances(
            embeddings[i].reshape(1, -1),
            embeddings[j].reshape(1, -1)
        )[0][0]

        pair_results.append({
            "sentence_1": i + 1,
            "sentence_2": j + 1,
            "similarity": similarity,
            "distance": distance
        })

        print(f"\nSentence {i + 1} ↔ Sentence {j + 1}")
        print(f"Similarity: {similarity:.4f}")
        print(f"Distance:   {distance:.4f}")
```

```python
print("RANKED SENTENCE PAIRS")
print("=" * 80)

for rank, result in enumerate(pair_results, start=1):

    i = result["sentence_1"] - 1
    j = result["sentence_2"] - 1

    print(f"\nRank {rank}")

    print("Sentence 1:")
    print(sentences[i])

    print("Sentence 2:")
    print(sentences[j])

    print(
        f"Similarity Score: "
        f"{result['similarity']:.4f}"
    )

    print(
        f"Cosine Distance: "
        f"{result['distance']:.4f}"
    )

    print("-" * 80)
```

```python
# only the highest pairs

best_pair = pair_results[0]

i = best_pair["sentence_1"] - 1
j = best_pair["sentence_2"] - 1

print("MOST SIMILAR PAIR")
print("=" * 60)

print("Sentence 1:")
print(sentences[i])

print("\nSentence 2:")
print(sentences[j])

print(
    "\nSimilarity Score:",
    round(best_pair["similarity"], 4)
)

print(
    "Cosine Distance:",
    round(best_pair["distance"], 4)
)
```
