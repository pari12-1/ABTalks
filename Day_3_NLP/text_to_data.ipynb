```python
import nltk
import string
import pandas as pd 

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
text = """
Natural Language Processing (NLP) is amazing!
It helps computers understand human language,
and it is used in chatbots, search engines, and more.
"""

# this is our raw text
```

```python
print("Original Text :")
print(text)
```

```python
# tokenization starts from here

tokens = word_tokenize(text)
print("Tokens :")
print(tokens)
```

```python
# converting is case-sensitive ... therefore converting it all to lowercase.

lower_tokens = [word.lower() for word in tokens]
print("Lower Tokens :")
print(lower_tokens)
```

```python
# removing punctuations

tokens_without_punctuation = [
    word
    for word in lower_tokens
    if word not in string.punctuation
]

print("After removing punctuations : ")
print(tokens_without_punctuation)
```

```python
# stop words

stop_words = set(stopwords.words("english"))
print(stop_words)
```

```python
# removing dtopwords from code

clean_tokens = [
    word
    for word in tokens_without_punctuation 
    if word not in stop_words
]
print("After removing stop words :")
print(clean_tokens)
```

```python
clean_text = " ".join(clean_tokens)

print("PreProcessed Text :")
print(clean_text)
```

```python
# converting text into BaW

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform([clean_text])
```

```python
feature_names = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(feature_names)
```

```python
print("Bag of Words:")
print(bow_matrix.toarray())
```

```python
bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=feature_names
)

bow_df
```

```python
# creting a reusable preprocessing functions 

def preprocess_text(text):

    # Tokenization
    tokens = word_tokenize(text)

    # Lowercase
    tokens = [
        word.lower()
        for word in tokens
    ]

    # Remove punctuation
    tokens = [
        word
        for word in tokens
        if word not in string.punctuation
    ]

    # Remove stop words
    stop_words = set(stopwords.words("english"))

    tokens = [
        word
        for word in tokens
        if word not in stop_words
    ]

    # Return cleaned text
    return " ".join(tokens)
```

```python
cleaned_dataset = [
    preprocess_text(text)
    for text in clean_tokens
]

print(cleaned_dataset)
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(cleaned_dataset)

bow_dataset = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

bow_dataset
```
